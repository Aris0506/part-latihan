# cleaning iranian dataset 10K transaction

```python
import pandas as pd

df = pd.read_csv('archive/trx-10k.csv')

print('========Lima baris pertama==============')
display(df.head())

print('\n----info tipe data----')
print(df.info())
```

**change data type**

```python
# pakai looping kalau mau hajar semua kolom teks jadi kategori secara otomatis
# Ambil semua kolom yang tipenya object/string, lalu ubah ke category
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype('category')

if 'time' in df.columns:
    df['time'] = pd.to_datetime(df['time'])

df.info()
```

**Handling missing values drop with threshold 5%**


```python
# cek missing values semua variable
print(df.isna().sum())
```

```python
threshold =  len(df) * 0.05
print(f'jumlah baris', threshold)

cols_to_drop = df.columns[df.isna().sum() <= threshold]
print(f'value kolom yang akan di drop', cols_to_drop)

df.dropna(subset=cols_to_drop, inplace=True)

print(df.isna().sum())
```
