```python
import pandas as pd

print('initial data:')
df = pd.read_csv('harvest_report.csv', index_col=0)
print(df)
df.info()
```

```python
# Kalau mau otomatis ngisi semua kolom teks yang kosong sebelum diubah jadi kategori:
# 1. Ambil semua kolom object
cols_object = df.select_dtypes(include=['object']).columns

# 2. Isi yang kosong dengan 'Missing' lalu ubah ke category
for col in cols_object:
    df[col] = df[col].fillna('Unknown').astype('category')

# 3. Untuk kolom angka, isi dengan 0 Hajar semua jenis angka (int & float)
cols_numeric = df.select_dtypes(include=['number']).columns
df[cols_numeric] = df[cols_numeric].fillna(0)

# 4. Konversi Tanggal (Manual karena formatnya spesifik)
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])


print(df.info())
```

**Kamu harus isi fillna() dulu baru astype('category'). Kalau sudah jadi kategori, nambahin nilai baru yang belum terdaftar di "kategori" itu bisa bikin error atau ribet menurut dokumentasi Pandas Fillna.**
